```python
def convert_time_to_frame(times, video_nums):
    frames = []
    for time, num in zip(times, video_nums):
        frame = (1000 * num) + ((time - (16.666 * num)) / 16.666) * 1000
        frames.append(int(frame))
    return frames
```

```python
def calculate_spike_times(trace, threshold):
    spike_times = {}

    for unit, activity in enumerate(trace):
        detect = False
        spikes = []
        for frame, val in enumerate(activity):
            if val > threshold and detect == False:
                detect = True
                spikes.append(frame)

            if val < threshold:
                detect = False

        spike_times[unit] = spikes

    return spike_times

def plot_2d_activity(ax, traces, title, start_eating_frames=None, stop_eating_frames=None):
    num_neurons, num_frames = traces.shape
    time_axis = np.arange(num_frames)
    summed_activity = traces.sum(axis=0)

    ax.plot(time_axis, summed_activity, label="Total Activity")
    ax.set_title(title)
    ax.set_xlabel("Frame Number")
    ax.set_ylabel("Total Fluorescence Activity")

    # Shade eating windows if provided
    if start_eating_frames is not None and stop_eating_frames is not None:
        for start, stop in zip(start_eating_frames, stop_eating_frames):
            ax.axvspan(start, stop, color='orange', alpha=0.2, label='Eating Period')

        # Avoid duplicate legend entries
        handles, labels = ax.get_legend_handles_labels()
        unique = dict(zip(labels, handles))
        ax.legend(unique.values(), unique.keys())

def plot_3d_activity(ax, traces, title):
    num_neurons, num_frames = traces.shape
    time_axis = np.arange(num_frames)
    for i in reversed(range(num_neurons)):
        y = np.full_like(time_axis, i)
        ax.plot(time_axis, y, traces[i])
    ax.set_title(title)
    ax.set_xlabel("Frame Number")
    ax.set_ylabel("Neuron Number")
    ax.set_zlabel("Fluorescence Value")

def plot_all(traces_exp, traces_ctrl):
    sns.set_theme(palette='flare')
    
    # Set up the figure
    fig = plt.figure(figsize=(20, 15))


    
    # 2D Experimental
    ax1 = fig.add_subplot(2, 2, 1)
    plot_2d_activity(ax1, traces_exp, "2D Experimental Activity Over Time")

    # 2D Control
    ax2 = fig.add_subplot(2, 2, 2)
    plot_2d_activity(ax2, traces_ctrl, "2D Control Activity Over Time")

    # 3D Experimental
    ax3 = fig.add_subplot(2, 2, 3, projection='3d')
    plot_3d_activity(ax3, traces_exp, "3D Experimental Activity Over Time")

    # 3D Control
    ax4 = fig.add_subplot(2, 2, 4, projection='3d')
    plot_3d_activity(ax4, traces_ctrl, "3D Control Activity Over Time")

    plt.tight_layout()
    plt.show()

def plot_single_neuron_comparison(traces_exp, traces_ctrl, unit):
    time_axis_exp = np.arange(traces_exp.shape[1])
    time_axis_ctrl = np.arange(traces_ctrl.shape[1])
    exp_activity = traces_exp[unit]
    ctrl_activity = traces_ctrl[unit]

    fig, ax = plt.subplots(2, 1, figsize=(10,10))



    ax[0].plot(time_axis_exp, exp_activity)
    ax[0].set_title(f"Neuron {unit} Activity Over Time: Experiment")
    ax[0].set_xlabel("Frame Number")
    ax[0].set_ylabel("Fluorescence Activity")

    ax[1].plot(time_axis_ctrl, ctrl_activity)
    ax[1].set_title(f"Neuron {unit} Activity Over Time: Control")
    ax[1].set_xlabel("Frame Number")
    ax[1].set_ylabel("Fluorescence Activity")

def plot_single_neuron_trace(traces, unit, start_eating_frames=None, stop_eating_frames=None):
    time_axis = np.arange(traces.shape[1])
    activity = traces[unit]

    plt.figure(figsize=(10, 5))
    plt.plot(time_axis, activity, label="Neuron Activity")
    plt.title(f"Neuron {unit} Activity Over Time")
    plt.xlabel("Frame Number")
    plt.ylabel("Fluorescence Activity")

    # Add shaded eating periods if provided
    if start_eating_frames is not None and stop_eating_frames is not None:
        for start, stop in zip(start_eating_frames, stop_eating_frames):
            plt.axvspan(start, stop, color='orange', alpha=0.2, label='Eating Period')

        # Avoid duplicate legend entries
        handles, labels = plt.gca().get_legend_handles_labels()
        unique = dict(zip(labels, handles))
        plt.legend(unique.values(), unique.keys())

    plt.tight_layout()
    plt.show()

def activity_3D (traces, experiment):
    num_neurons = traces.shape[0]
    num_frames = traces.shape[1]

    sns.set_theme(palette='flare')

    fig = plt.figure(figsize = (13, 13))
    ax = fig.add_subplot(111, projection='3d')

    time_axis = np.arange(traces.shape[1])
    for i in reversed(range(num_neurons)):
        y = np.full_like(time_axis, i)
        ax.plot(time_axis, y, traces[i])
        
        
        
    ax.set_title(f"Fluoresence Activity Over Time Per Neuron: {experiment}")
    ax.set_xlabel('Frame Number')
    ax.set_ylabel('Neuron Number')
    ax.set_zlabel('Fluorescence Value')
    plt.show()
    
def activity_2D (traces, experiment):
    
    num_neurons = traces.shape[0]
    num_frames = traces.shape[1]
    summed_activity = []



    fig, ax = plt.subplots(figsize = (10, 10))
    time_axis = np.arange(traces.shape[1])

    for i in time_axis:
        sum_activity = 0
        for j in range(num_neurons):
            sum_activity += traces[j][i]
        summed_activity.append(sum_activity)
        
    ax.plot(time_axis, summed_activity)
    plt.title(f"Fluoresence Activity Over Time: {experiment}")
    ax.set_xlabel('Frame Number')
    ax.set_ylabel('Total Fluorescence activity')
    plt.show()
    
def compare_activity(traces_exp, traces_ctrl):
    summed_activity_exp = []
    summed_activity_ctrl = []
    max_frames = min(traces_exp.shape[1], traces_ctrl.shape[1])
    num_neurons = traces_exp.shape[0]
    time_axis = np.arange(max_frames)
    for i in time_axis:
        sum_activity_exp = 0
        sum_activity_ctrl = 0
        for j in range(num_neurons):
            sum_activity_exp += traces_exp[j][i]
            sum_activity_ctrl += traces_ctrl[j][i]
            
        summed_activity_exp.append(sum_activity_exp)
        summed_activity_ctrl.append(sum_activity_ctrl)
    
    
    plt.bar(['experimental', 'control'], [sum(summed_activity_exp), sum(summed_activity_ctrl)])
    plt.title('Total Activity Control vs Experimental')
    plt.xlabel('Condition')
    plt.ylabel('Total Fluorescence activity')
    plt.show()

'''
def plot_rasters(spike_times, experiment):
    fig = plt.figure(figsize=(20,10))

    for unit in spike_times:
        spikes = spike_times[unit]
        for spike in spikes:
            plt.vlines(spike, unit + 0.5, unit + 1.5, color='black')

    plt.xlabel('Frame Number')
    plt.ylabel('Neuron Number')

    plt.title(f"Raster Plot of Each Neuron Across Time {experiment}")
    plt.ylim(0.5, len(spike_times) + 0.5)
    plt.legend()
    plt.show()
'''

def plot_rasters(spike_times, experiment, bin_size=30):
    all_spikes = [spike for spikes in spike_times.values() for spike in spikes]

    
    max_time = int(max(all_spikes)) + 1
    num_bins = (max_time + bin_size - 1) // bin_size  
    binned_spikes = np.zeros(num_bins)

    # Count spikes in each bin
    for spikes in spike_times.values():
        for spike in spikes:
            bin_index = int(spike) // bin_size
            if bin_index < num_bins:
                binned_spikes[bin_index] += 1

    bin_edges = np.arange(0, num_bins * bin_size, bin_size)
    bin_centers = bin_edges + bin_size / 2

    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(20, 12), height_ratios=[3, 1], sharex=False)


    for unit in spike_times:
        spikes = spike_times[unit]
        for spike in spikes:
            ax1.vlines(spike, unit + 0.5, unit + 1.5, color='black')

    ax1.set_ylabel('Neuron Number')
    ax1.set_title(f"Raster Plot of Each Neuron Across Time: {experiment}")
    ax1.set_ylim(0.5, len(spike_times) + 0.5)

    ax2.plot(bin_centers, binned_spikes, marker='o', linestyle='-', color='blue')
    ax2.set_xlabel('Frame Number')
    ax2.set_ylabel('Total Spikes')
    ax2.set_title(f'Summed Spike Activity Across Neurons (Binned Every {bin_size} Frames)')

    plt.tight_layout()
    plt.show()

def plot_inferred_spikes(S, percentile=99, neuron_label="Neuron", frame_rate=None, color='tomato'):
    """
    Plot raw inferred spike trace and thresholded binary spikes.

    Parameters:
    - S : np.ndarray
        Deconvolved spike activity (estimates.S[n, :] for one neuron).
    - percentile : float
        Percentile to use for thresholding (e.g., 95).
    - neuron_label : str
        Label for plot title.
    - frame_rate : float or None
        If given, x-axis will be converted to seconds (time = frames / frame_rate).
    - color : str
        Color for plots.
    """

    
    S = np.asarray(S)
    threshold = np.percentile(S, percentile)
    print(threshold)
    binary_spikes = (S > threshold).astype(int)
    spike_times = np.where(binary_spikes == 1)[0]

    # Time axis
    if frame_rate is not None:
        time = np.arange(len(S)) / frame_rate
        x_label = "Time (s)"
    else:
        time = np.arange(len(S))
        x_label = "Frame Number"

    plt.figure(figsize=(15, 5))
    
    # Raw spike activity
    plt.plot(time, S, label='Inferred Activity (S)', color=color, alpha=0.6)
    plt.axhline(threshold, color='gray', linestyle='--', label=f'{percentile}th Percentile Threshold')

    # Tick marks for spikes
    plt.vlines(time[spike_times], ymin=0, ymax=threshold*0.9, color='black', linewidth=1.0, alpha=0.7, label='Thresholded Spikes')

    plt.title(f"{neuron_label} — Inferred Spikes with {percentile}th Percentile Threshold")
    plt.xlabel(x_label)
    plt.ylabel("Inferred Spike Activity (S)")
    plt.legend()
    plt.tight_layout()
    plt.show()

def threshold_S(S, percentile=99.9):
    spike_trains = {}

    for neuron, spike_train in enumerate(S):
        threshold = np.percentile(spike_train, percentile)
        binary_spikes = (spike_train > threshold).astype(int)
        spike_times = np.where(binary_spikes == 1)[0]
        spike_trains[neuron] = spike_times

    return spike_trains

def compute_average_f(spike_times, calcium_traces):
    avg_amplitudes = {}
    for neuron, spikes in spike_times.items():

        if len(spikes) == 0:
            avg_amplitudes[neuron] = 0
            continue

        valid_spikes = spikes[(spikes >= 0) & (spikes < calcium_traces.shape[1])]
        if len(valid_spikes) == 0:
            avg_amplitudes[neuron] = np.nan
            continue

        trace = calcium_traces[neuron]
        avg_amplitudes[neuron] = np.mean(trace[valid_spikes])

    return avg_amplitudes

def compare_amplitudes_boxplot(experimental_amplitudes, control_amplitudes, show_individual_points=True):
    """
    Create a box plot comparing calcium event amplitudes and run a t-test.

    Parameters:
    - control_amplitudes : list or array of float
    - experimental_amplitudes : list or array of float
    - show_individual_points : bool, whether to overlay raw data points
    """
    # Clean NaNs if present
    control_amplitudes = np.array(control_amplitudes)
    experimental_amplitudes = np.array(experimental_amplitudes)
    control_amplitudes = control_amplitudes[~np.isnan(control_amplitudes)]
    experimental_amplitudes = experimental_amplitudes[~np.isnan(experimental_amplitudes)]

    # T-test
    t_stat, p_val = ttest_ind(control_amplitudes, experimental_amplitudes, equal_var=False)
    print(f"Two-tailed Student's t-test p-value: {p_val:.4f}")

    # Plot
    fig, ax = plt.subplots(figsize=(6, 6))
    data = [control_amplitudes, experimental_amplitudes]
    labels = ['Control', 'Experimental']
    boxprops = dict(linewidth=1.5)
    medianprops = dict(linewidth=2.5, color='black')

    ax.boxplot(data, labels=labels, boxprops=boxprops, medianprops=medianprops, showfliers=False)

    if show_individual_points:
        # Add jittered scatter
        for i, group in enumerate(data):
            x = np.random.normal(i+1, 0.05, size=len(group))
            ax.plot(x, group, 'k.', alpha=0.7)

    ax.set_ylabel("Calcium Event Amplitude (ΔF/F)")
    ax.set_title(f"Comparison of Event Amplitudes, p={p_val:.2f}, t={abs(t_stat):.2f}")
    


    plt.tight_layout()
    plt.show()

def compare_amplitudes_with_matches(experimental_amplitudes, control_amplitudes, assignments, show_individual_points=True):
    """
    Create a box plot comparing calcium event amplitudes and run a t-test.
    Unmatched neurons (based on NaNs in `assignments`) are plotted in red.

    Parameters:
    - experimental_amplitudes : list or array of float
    - control_amplitudes : list or array of float
    - assignments : np.ndarray of shape (n_pairs, 2) where NaNs indicate unmatched neurons
    - show_individual_points : bool, whether to overlay raw data points
    """
    # Convert and clean NaNs
    control_amplitudes = np.array(control_amplitudes)
    experimental_amplitudes = np.array(experimental_amplitudes)
    control_amplitudes = control_amplitudes[~np.isnan(control_amplitudes)]
    experimental_amplitudes = experimental_amplitudes[~np.isnan(experimental_amplitudes)]

    # T-test
    t_stat, p_val = ttest_ind(control_amplitudes, experimental_amplitudes, equal_var=False)
    print(f"Two-tailed Student's t-test p-value: {p_val:.4f}")

    # Plot
    fig, ax = plt.subplots(figsize=(6, 6))
    data = [control_amplitudes, experimental_amplitudes]
    labels = ['Control', 'Experimental']
    boxprops = dict(linewidth=1.5)
    medianprops = dict(linewidth=2.5, color='black')

    ax.boxplot(data, labels=labels, boxprops=boxprops, medianprops=medianprops, showfliers=False)

    if show_individual_points:
        # Add jittered scatter
        matched_experimental = set(assignments[:, 0][~np.isnan(assignments[:, 1])].astype(int))
        matched_control = set(assignments[:, 1][~np.isnan(assignments[:, 0])].astype(int))
        print(matched_control)

        for i, group in enumerate(data):
            x = np.random.normal(i + 1, 0.05, size=len(group))
            for j, val in enumerate(group):
                if i == 0:  # Control
                    color = 'red' if j not in matched_control else 'black'
                else:       # Experimental
                    color = 'red' if j not in matched_experimental else 'black'
                ax.plot(x[j], val, '.', color=color, alpha=0.7)

    ax.set_ylabel("Calcium Event Amplitude (ΔF/F)")
    ax.set_title(f"Comparison of Event Amplitudes, p={p_val:.2f}, t={abs(t_stat):.2f}")

    plt.tight_layout()
    plt.show()

def compute_and_plot_peth(traces, event_times, window=15, fps=15):
    """
    Compute and plot a peri-event time histogram (PETH) for a 1D trace around given event times.

    Parameters:
    - traces: list or array of activity values (2D)
    - event_times: list of int, indices of event frames (e.g., eating stops)
    - window: int, number of frames before and after the event
    - fps: frames per second (for labeling time axis)
    """

    snippets = []
    for trace in traces:
        # Extract event-aligned snippets
        
        for event in event_times:
            start = event - window
            stop = event + window
            if start >= 0 and stop < len(trace):
                snippet = trace[start:stop]
                snippets.append(snippet)

    if not snippets:
        print("No valid events within bounds.")
        return

    # Compute mean and SEM
    n_events = len(snippets)
    n_timepoints = len(snippets[0])
    mean_trace = [0.0] * n_timepoints
    sem_trace = [0.0] * n_timepoints

    for t in range(n_timepoints):
        values = [s[t] for s in snippets]
        mean = sum(values) / n_events
        mean_trace[t] = mean
        var = sum((v - mean) ** 2 for v in values) / n_events
        std = var ** 0.5
        sem_trace[t] = std / (n_events ** 0.5)

    # Time axis in seconds
    time_axis = [(i - window) / fps for i in range(n_timepoints)]

    # Plot
    plt.figure(figsize=(8, 4))
    plt.plot(time_axis, mean_trace, label="Mean Activity")
    plt.fill_between(time_axis,
                     [m - s for m, s in zip(mean_trace, sem_trace)],
                     [m + s for m, s in zip(mean_trace, sem_trace)],
                     alpha=0.3)
    plt.axvline(0, color='red', linestyle='--', label='Event (Eating Cessation)')
    plt.xlabel("Time (s) Relative to Eating Cessation")
    plt.ylabel("Activity (Fluorescence)")
    plt.title("Mean Activity and SEM Alligned to Eating Cessation")
    plt.legend()
    plt.tight_layout()
    plt.show()

def plot_normalized_population_heatmap(calcium_traces, event_frames, window=(-50, 100), fps=15):
    """
    Plot a stacked heatmap of normalized peri-event activity for all neurons.

    Parameters:
    - calcium_traces: ndarray of shape (num_neurons, num_frames)
    - event_frames: list or array of event onset frame indices
    - window: tuple (pre_event, post_event) in frames
    - fps: frames per second, for optional time labeling
    """
    num_neurons, num_frames = calcium_traces.shape
    pre, post = window
    peri_length = post - pre
    all_trials = []

    for frame in event_frames:
        start = frame + pre
        end = frame + post
        if start < 0 or end > num_frames:
            continue  # skip events too close to edges
        segment = calcium_traces[:, start:end]
        print(segment.shape)
        all_trials.append(segment)


    if not all_trials:
        print("No valid events within the specified window.")
        return

    # Average across events → shape (num_neurons, peri_length)
    avg_response = np.mean(all_trials, axis=0)
    print(avg_response.shape)
    # Normalize each neuron (z-score)
    norm_response = (avg_response - np.mean(avg_response, axis=1, keepdims=True)) / \
                    (np.std(avg_response, axis=1, keepdims=True) + 1e-6)  # avoid divide by 0

    # Plot heatmap
    plt.figure(figsize=(10, 8))
    plt.imshow(avg_response, aspect='auto', cmap='flare',
               extent=[pre/fps, post/fps, 0, num_neurons])
    plt.colorbar(label='Averaged Activity')
    plt.xlabel("Time from Eating Cessation (s)")
    plt.ylabel("Neuron Index")
    plt.title("Neuronal Activity Aligned to Eating Cessation")
    plt.axvline(0, color='red', linestyle='--', label='Eating Cessation')
    plt.legend()
    plt.tight_layout()
    plt.show()

def plot_normalized_population_heatmap_full_recording(calcium_traces, start_eating_frames, stop_eating_frames, fps=15):
    num_neurons, num_frames = calcium_traces.shape

    # Plot heatmap
    fig, ax = plt.subplots(figsize=(16, 8))

    im = ax.imshow(calcium_traces, aspect='auto', cmap='flare')
    
    # Add shaded eating periods if provided
    legend_entries = []
    if start_eating_frames is not None and stop_eating_frames is not None:
        num_periods = len(start_eating_frames)
        colors = cmap.get_cmap('tab10', num_periods)  # or 'tab20', 'hsv', etc.

        for i, (start, stop) in enumerate(zip(start_eating_frames, stop_eating_frames)):
            ax.axvspan(start, stop, color=colors(i), alpha=0.3, label=f'Eating Period {i+1}')


    # Colorbar (placed inside to avoid clash with external legend)
    cbar = fig.colorbar(im, ax=ax)
    cbar.set_label('Activity')

    # Labels and title
    ax.set_xlabel("Frame #")
    ax.set_ylabel("Neuron Index")
    ax.set_title("Neuronal Heatmaps")
    ax.invert_yaxis()


    plt.show()


```
